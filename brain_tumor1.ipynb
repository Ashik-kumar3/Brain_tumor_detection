```python
#Preprocessing

#Input and output seperation

xtrain = []
ytrain= []
datadir = r'C:\Users\ASHIK KUMAR\Desktop\AI\AI\DL\CNN\Training'
categories = ['glioma', 'meningioma', 'notumor', 'pituitary']
```

```python
for i in categories:
    path = os.path.join(datadir,i)
    for img in os.listdir(path):
        img_arr = imread(os.path.join(path,img))
        img_res = resize(img_arr,(150,150,1))
        xtrain.append(img_res)
        ytrain.append(categories.index(i))
```

```python
xtest=[]
ytest=[]
datadir1=r'C:\Users\ASHIK KUMAR\Desktop\AI\AI\DL\CNN\Testing'
categories1 = ['glioma', 'meningioma', 'notumor', 'pituitary']
```

```python
for j in categories1:
    path1 = os.path.join(datadir1,j)
    for img1 in os.listdir(path1):
        img_arr1 = imread(os.path.join(path1,img1))
        img_res1 = resize(img_arr1,(150,150,1))
        xtest.append(img_res1)
        ytest.append(categories1.index(j))
```

```python
xtrain = np.array(xtrain)
xtest = np.array(xtest)
ytrain = np.array(ytrain)
ytest = np.array(ytest)
```

```python
#Model Building
from tensorflow.keras.models import Sequential
model = Sequential()
```

```python
#Adding layers
model.add(Conv2D(3,(3,3),activation='relu',input_shape=(150,150,1)))
model.add(MaxPool2D(2,2))
model.add(Flatten())
model.add(Dense(10,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(4,activation='softmax'))
```

```python
#compilation
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
#model fitting
model.fit(xtrain,ytrain,epochs=10,batch_size=200)
```

```python
#model evaluation
model.evaluate(xtest,ytest,batch_size=200)
```
